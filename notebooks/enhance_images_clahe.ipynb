```python
import os
import cv2
import pandas as pd
from tqdm import tqdm

# Paths
root_dir = "/home/sergio/datasets/laticifers"
gray_dir = os.path.join(root_dir, "gray_images")
clahe_dir = os.path.join(root_dir, "clahe_images")
os.makedirs(clahe_dir, exist_ok=True)

# Load dataset index
df = pd.read_csv(os.path.join(root_dir, "laticifer_dataset_index.csv"))

# Apply CLAHE and save processed images
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))

clahe_paths = []

for _, row in tqdm(df.iterrows(), total=len(df), desc="Applying CLAHE"):
    gray_path = os.path.join(root_dir, row["gray_image_path"])
    img = cv2.imread(gray_path, cv2.IMREAD_GRAYSCALE)

    if img is None:
        clahe_paths.append("")
        continue

    img_clahe = clahe.apply(img)

    save_path = os.path.join(clahe_dir, os.path.basename(gray_path))
    cv2.imwrite(save_path, img_clahe)

    # Save relative path
    rel_path = os.path.relpath(save_path, root_dir)
    clahe_paths.append(rel_path)

# Add new column to dataframe
df["clahe_image_path"] = clahe_paths

# Save updated CSV
df.to_csv(os.path.join(root_dir, "laticifer_dataset_index.csv"), index=False)
print("✅ CLAHE preprocessing complete and CSV updated.")
```
